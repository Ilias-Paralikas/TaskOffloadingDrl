```python
import os
import shutil

def delete_subfolders_named_x(folder_path,dirname):
    for root, dirs, files in os.walk(folder_path, topdown=False):
        for dir_name in dirs:
            if dir_name == dirname:
                dir_path = os.path.join(root, dir_name)
                shutil.rmtree(dir_path)
                print(f"Deleted folder: {dir_path}")

# Example usage
folder_path = 'meta_plots/logs'
dirname = 'actions'
delete_subfolders_named_x(folder_path,dirname)
```
